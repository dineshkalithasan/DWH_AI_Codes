# Numbers & mathematical logic
Prime numbers, Fibonacci series, factorials, Armstrong numbers, and other frequently asked logic problems.

```python
#prime number check
num = int(input("Enter a number: "))
if num > 1:
    for i in range(2, int(num/2)+1):
        if (num % i) == 0:
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
```

```python
#fibonacci check
fib = int(input("Enter the number of terms: "))
f1 = int(input("Enter the first term: "))
f2 = int(input("Enter the second term: "))
print("Fibonacci sequence:")
print(f"{f1}\n{f2}")

for i in range(fib):
    f3=f1+f2
    f1=f2
    f2=f3
    print(f3)
```

```python
#amstrong number 
num = int(input("Enter a number: "))
sum = 0
temp = num
while temp > 0:
   digit = temp % 10
   sum += digit ** 3
   temp //= 10
if num == sum:
   print(num,"is an Armstrong number")
else:
   print(num,"is not an Armstrong number")
```
