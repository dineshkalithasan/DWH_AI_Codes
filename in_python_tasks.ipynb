### F. Type identification & Casting
Use Case 1:
Sample code: As we didn’t covered this function input() yet in our session, I am giving the below sample code for your reference.
Write a program that asks for an employee’s age.

1. Checks its type is of string
print(type(age))
2. Converts it to int (continue writing your program from here..)
3. Prints the years pending for retirement, for eg. 60 is the retirement age.

```python
age = input("Enter your age: ")
print("type of variable age is:", type(age))
age = int(age)
print("type of variable age is converted to :", type(age))

retirement_age = 60
age_left_for_retirement = (retirement_age - age)

if age_left_for_retirement > 0:
    print(f"Your age is: {age} and you have {age_left_for_retirement} years left for your retirement ")
else:
    print(f"Your age is: {age} and you have already retired age")
```
