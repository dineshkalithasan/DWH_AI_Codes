### G. Data types and casting
Use Case 1 — Employee Salary Breakdown Using Numeric & String Types
Employee Salary Breakdown

a. Write a program that asks the user for: employee_name (string).
base_salary (float). 
hra_percent (integer). 
bonus_amount (float).

B. Convert inputs to the correct datatype if required.
Calculate:
 HRA = base_salary * (hra_percent / 100)
 Total Salary = base_salary + HRA + bonus_amount

C. Print the output 

```python
#a
employee_name = input("Enter employee name: ")
base_salary = float(input("Enter base salary: "))
hra_percent = int(input("Enter HRA percentage: "))
bonus_amount = float(input("Enter bonus amount: "))

#b
hra_amount = (base_salary * hra_percent) / 100
total_salary = base_salary + hra_amount + bonus_amount

#c
print(f"Employee Name: {employee_name}")
print(f"Base Salary: {base_salary}")
print(f"HRA Percentage: {hra_percent}")
print(f"Bonus Amount: {bonus_amount}")
print(f"HRA Amount: {hra_amount}") 
print(f"Total Salary: {total_salary}")
```

Use Case 2: 
Student Result Classification<br>
a. Write a program that takes marks as input (initially as a string). <br>
B. Check if the value can be converted to float.<br>
C. Then classify (try using if condition with the help of AI, however we will learn about if condition soon):
Marks >= 90 --> Outstanding
 Marks >= 75 --> Excellent
 Marks >= 50 --> Pass
 Marks < 50 --> Fail<br>
D. If the input is not numeric, print:
 Invalid marks entered — Please provide numeric input.


```python
#Student Result Classification<br>
#Write a program that takes marks as input (initially as a string). <br>
marks = input("Enter marks: ")

#Check if the value can be converted to float.
try:
    marks = float(marks)
#Then classify (try using if condition with the help of AI, however we will learn about if condition soon):
    if marks < 0 or marks > 100:
        print("Invalid marks entered. Please enter a value between 0 and 100.")
    elif marks >= 90:
        print("Outstanding")
    elif marks >= 75:
        print("Excellent")
    elif marks >= 50:
        print("Pass")
    else:
        print("Fail")
#If the input is not numeric, print:
except ValueError:
    print("Invalid marks entered — Please provide numeric input.")
```
